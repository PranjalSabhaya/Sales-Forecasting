```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns',50)
pd.set_option('display.width',120)
```

```python
DATA_PATH = "../data/raw/"
INTERIM_PATH = "../data/interim/"

calendar_df = pd.read_csv(DATA_PATH + "calendar.csv")
sales_val_df = pd.read_csv(DATA_PATH + "sales_train_validation.csv")
```

```python
sales_val_df.columns
```

```python
id_columns = ['id', 'item_id', 'dept_id', 'cat_id', 'store_id', 'state_id']

day_columns = [col for col in sales_val_df.columns if col.startswith('d_')]

print('Day columns:',len(day_columns))
```

```python
sales_long_df = sales_val_df.melt(
    id_vars=id_columns,
    value_vars=day_columns,
    var_name="d",
    value_name="sales"
)

sales_long_df.head()
```

```python
calendar_df.columns
```

```python
calendar_small_df = calendar_df[calendar_df.columns]
```

```python
sales_long_df = sales_long_df.merge(
    calendar_small_df,
    on="d",
    how="left"
)
```

```python
sales_long_df.head()
```

```python
sales_long_df["date"] = pd.to_datetime(sales_long_df["date"])
```

```python
sales_long_df["sales"].isna().sum()
```

Sorting is performed later at the item–store level during feature engineering.
